```python
import numpy as np
import matplotlib.pyplot as plt

T = 300
t = np.linspace(0, T, 300)
Acc = 0.9 * t / T  # Linear accuracy growth
beta = 0.03

reward = (1 - Acc) * (1 - np.exp(-beta * t))

plt.figure(figsize=(10, 5))
plt.plot(t, reward, label='Reward', color='blue')
plt.plot(t, Acc, label='Accuracy', color='green', linestyle='--')
plt.xlabel('Iteration (t)')
plt.ylabel('Reward / Accuracy')
plt.title('Reward Function vs. Accuracy Over Time')
plt.legend()
plt.grid(True)
plt.show()
```
